# 수급 보유 변화 탐색 (KIS holdings)

이 노트북은 `dashboard/holdings_chart.py` 의 `load_holdings` / `make_figure` 를 import 해서 임의 종목·기간을 빠르게 그려본다.

차트 구성 (3 패널):

1. **종가 (line) + 가격 변동률 (bar)** — 양수 빨강 / 음수 파랑
2. **11 주체 누적 보유량 (cum_qty)** — 외국인(통합/등록/비등록) · 기관 6 sub · 개인 · 기타법인
3. **10 sub 보유 비율 (pct)** — `pct_mode="cumulative"` (디폴트, 누적) 또는 `"daily"` (그날의 net_qty 분모)

**해석 팁**
- Row 2 의 라인 기울기가 가파를수록 그 주체의 매수/매도 강도가 강함
- 첫 거래일 근처는 누적이 작아 row 3 비율이 0% 근방에서 흔들리는 게 정상
- 범례를 카테고리(외국인/기관/개인/기타법인)별로 클릭하면 그룹 한 번에 토글

**조작법** — 마우스 스크롤 = 호버 중인 패널 확대/축소 · 드래그 = 박스 줌 · 더블클릭 = 리셋


```
# 의존성 + 함수 import
from dashboard.holdings_chart import (
    load_holdings, make_figure, PLOTLY_CONFIG, SUBJECT_LABELS,
)

# plotly 인라인 렌더링
import plotly.io as pio
pio.renderers.default = "notebook_connected"
```

```
# 삼성전자 — 전체 기간 (누적 비중)
df, meta = load_holdings("005930")
print(f"{meta.get('company_name')} ({meta.get('ticker')}) · {meta.get('market')}")
print(f"기간: {df['date'].iloc[0]} ~ {df['date'].iloc[-1]} · 총 {len(df)}일")
fig = make_figure(df, title=f"{meta.get('company_name')} ({meta.get('ticker')}) — 누적")
fig.show(config=PLOTLY_CONFIG)
```

```
# 같은 종목 — 일별 비중 모드
fig = make_figure(
    df,
    title=f"{meta.get('company_name')} ({meta.get('ticker')}) — 일별",
    pct_mode="daily",
)
fig.show(config=PLOTLY_CONFIG)
# 차이점: row 3 면적이 매일 출렁임. 거래가 없는 주체는 즉시 0%.
```

```
# SK하이닉스 — 전체 기간
df, meta = load_holdings("000660")
print(f"{meta.get('company_name')} · {len(df)}일")
fig = make_figure(df, title=f"{meta.get('company_name')} ({meta.get('ticker')})")
fig.show(config=PLOTLY_CONFIG)
```

```
# 좁은 기간 — 최근 1년만 보기
df, meta = load_holdings("005930", start="2025-01-01")
print(f"{meta.get('company_name')} · 1년 · {len(df)}일")
fig = make_figure(df, title=f"{meta.get('company_name')} 2025-01-01 ~")
fig.show(config=PLOTLY_CONFIG)
```

```
# 일부 주체만 표시 — 외국인 vs 개인만 강조
df, meta = load_holdings("005930", start="2025-01-01")
fig = make_figure(
    df,
    title="외국인 vs 개인 — 2025",
    visible_subjects=["foreign", "retail"],
)
fig.show(config=PLOTLY_CONFIG)
# 나머지 9 주체는 범례에서 'legendonly' 상태 — 클릭하면 다시 표시
```
